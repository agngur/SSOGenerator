```python
import numpy as np
import matplotlib.pyplot as plt

# Generate angles from 0 to 2π
theta = np.linspace(0, 2 * np.pi, 100)

# Fill circle (Star) with colormap depending on its luminosity
#TOBEDONE

# Define radius and center
radius = 1
x_center, y_center = 0, 0

# Calculate x and y coordinates
x = x_center + radius * np.cos(theta)
y = y_center + radius * np.sin(theta)

# Plot the circle
plt.plot(x, y)
plt.gca().set_aspect('equal') # Ensure aspect ratio is equal
plt.title("Circle Using Parametric Equation")
plt.show()
```

```python
# 2digit of year + Day Of Year 
tle[1].split()[3]

from datetime import timedelta

# first object in space was deployed in 1957 (1957-001A)
# there will be no TLEs older than that (for sure)
_year = tle[1].split()[3][:2]
doy = tle[1].split()[3][2:]
year = "19"+_year if int(_year) >= 57 else "20"+_year 

print(year, doy) 
# so we get 1 of January of a given year and then add DOYs-1
tle_date = datetime(int(year), 1, 1) + timedelta(float(doy) - 1)
tle_age = (datetime.today() - tle_date).total_seconds() / (3600*24) # days

print("TLE epoch is", tle_date.isoformat(), end=" ")
print("and it is %.2f days old" % tle_age)
```
